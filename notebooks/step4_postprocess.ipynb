```python
"""Transform the predictions from the normalized data back to the original CTA data space using the transforms from data.py. This is necessary because the predictions are made on normalized data, and we want to visualize them in the original CTA space."""

import sys
import os
sys.path.append('..')

from src.processing import transform_all, CONSTANTS

root_pred = ".."


root_pred_val = f"../{CONSTANTS.NORM_PRED_PATH}"
root_ncct_val = f"../{CONSTANTS.DATA_PATH}/imagesTr"
models_val = ["cknn", "knn", "durag"]

root_pred_ts = f"../{CONSTANTS.TEST_PRED_PATH}"
root_ncct_ts = f"../{CONSTANTS.DATA_PATH}/imagesTs"
models_ts = ["durag"]



for (root_pred, root_ncct, models) in [(root_pred_val, root_ncct_val, models_val), (root_pred_ts, root_ncct_ts, models_ts)]:
    for model in models:
        root_pred_model = f"{root_pred}/images_norm_{model}/"
        root_out_model = f"{root_pred}/images_{model}/"
        
        os.makedirs(root_out_model, exist_ok=True)
        
        transform_all(root_pred_model, root_ncct, root_out_model, to_int=True)
        print(f"end=transform_all for {model} in {root_pred}")
```
